```python
import mediapipe as mp
import cv2 
import time
import json
from datetime import datetime
```

```python
"""Setup the camera and the MediaPipe Hands model."""
mp_hands = mp.solutions.hands
hands = mp_hands.Hands(min_detection_confidence=0.7, min_tracking_confidence=0.5, max_num_hands=2)
mp_drawing = mp.solutions.drawing_utils


camera_index = 0
cap = cv2.VideoCapture(camera_index)

fps = cap.get(cv2.CAP_PROP_FPS)
print(f"Frames per second: {fps}")

prev_frame_time = 0
new_frame_time = 0

isRecording = False

sign = "rotate-left"
data = []
seq = 0
while cap.isOpened():
    success, image = cap.read()
    if not success:
        print("Ignoring empty camera frame.")
        continue

    image = cv2.cvtColor(cv2.flip(image, 1), cv2.COLOR_BGR2RGB)
    image.flags.writeable = False

    results = hands.process(image)
    image.flags.writeable = True
    image = cv2.cvtColor(image, cv2.COLOR_RGB2BGR)

    # If any hand is detected
    if results.multi_hand_landmarks:
        for hand_landmarks, handedness in zip(results.multi_hand_landmarks, results.multi_handedness):
            mp_drawing.draw_landmarks(
                image, hand_landmarks, mp_hands.HAND_CONNECTIONS)
            hand_data = []
            hand_label = handedness.classification[0].label
            for id, lm in enumerate(hand_landmarks.landmark):                
                hand_data.append({"id": id, "x": lm.x, "y": lm.y, "z": lm.z})
            data.append({"seq":  seq, "hand_landmarks": hand_data, "hand_label": hand_label})
            seq += 1            

    # Rewrite the FPS for development purposes.
    new_frame_time = time.time()
    actual_fps = 1 / (new_frame_time - prev_frame_time)
    prev_frame_time = new_frame_time

    cv2.putText(image, f'FPS: {int(actual_fps)}', (10, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2, cv2.LINE_AA)
    cv2.putText(image, "Recording" if isRecording else "Recording stopped", (10, 100), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2, cv2.LINE_AA)
    cv2.putText(image, sign, (200, 30), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2, cv2.LINE_AA)

    # Display the image.
    cv2.imshow('MediaPipe Hands', image)
    

    key = cv2.waitKey(5) & 0xFF


    if key == ord(' '):
        if isRecording: 
            path = f"data/{sign}/{datetime.now().isoformat().replace(":","-")}-{sign}.json"
            with open (path, "w") as f:
                json.dump(data, f, indent=4)
            data = []
            seq = 0
        isRecording = not isRecording
    elif key == ord('t'):
        sign = "rotate-left"
    elif key == ord('y'):
        sign = "rotate-right"
    elif key == ord('u'):
        sign = "zoom-in"
    elif key == ord('i'):
        sign = "zoom-out"
    elif key == ord('o'):
        sign = "slide-left"
    elif key == ord('p'):
        sign = "slide-right"
    elif key == 27:
        break
        


cap.release()
cv2.destroyAllWindows()

print("Stopped recording.")
```
